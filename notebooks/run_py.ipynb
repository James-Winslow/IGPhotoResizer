```python
from PIL import Image, ImageDraw
import random

# Generate a complex synthetic image with texture
random.seed(42)
width, height = 731, 943
img = Image.new("RGB", (width, height), (100, 150, 200))
draw = ImageDraw.Draw(img)

for _ in range(50):
    x0, y0 = random.randint(0, width), random.randint(0, height)
    x1, y1 = x0 + random.randint(20, 150), y0 + random.randint(20, 150)
    draw.ellipse([x0, y0, x1, y1], fill=(random.randint(0,255), random.randint(0,255), random.randint(0,255)))
    draw.rectangle([x0//2, y0//2, x1//2+200, y1//2+200], fill=(random.randint(0,255), random.randint(0,255), random.randint(0,255)))

padding_result = padding_resize(img, 1080, 1080)
simple_result  = simple_resize(img, 1080, 1080)

print("\nPadding metrics:")
print(compute_all_metrics(img, padding_result, "padding_resize", "complex_test"))
print("\nSimple metrics:")
print(compute_all_metrics(img, simple_result, "simple_resize", "complex_test"))
```
